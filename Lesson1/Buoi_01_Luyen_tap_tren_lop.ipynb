## 1 · Làm quen với biến và kiểu dữ liệu 
Đọc ví dụ: tên biến giúp ta gọi lại một giá trị. `type()` cho biết kiểu dữ liệu.

````python
course_name = "Python"
lesson_number = 1
print(course_name, lesson_number)
print(type(course_name))
print(type(lesson_number))
````

**Em làm:** tạo bốn biến theo yêu cầu. Sau đó dùng `print()` và `type()` để xem giá trị và kiểu của từng biến.

| Biến | Giá trị | Kiểu cần dùng |
|---|---|---|
| `student_name` | Một tên bất kỳ | `str` |
| `student_age` | 18 | `int` |
| `practice_score` | 8.5 | `float` |
| `is_ready` | Đúng | `bool` |

Gợi ý: giá trị Boolean của Python được viết là `True` hoặc `False`.

````python
student_name = None
student_age = None
practice_score = None
is_ready = None

# Viết thêm các lệnh print() và type() tại đây.
````

````python
# Ô kiểm tra có sẵn — không cần sửa hoặc học cú pháp trong ô này.
if student_name is None or student_age is None or practice_score is None or is_ready is None:
    print("✏️ Hãy hoàn thành các vị trí None ở bài phía trên.")
else:
    assert type(student_name) is str and len(student_name.strip()) > 0 and type(student_age) is int and student_age == 18 and type(practice_score) is float and practice_score == 8.5 and type(is_ready) is bool and is_ready is True, "Chưa đúng. Hãy xem lại yêu cầu và chạy lại ô bài làm."
    print("✅ Kết quả đúng!")
````

## 2 · Máy tính mini 
Cho `a = 17`, `b = 5`. Điền **biểu thức sử dụng hai biến** vào các chỗ trống, thay vì ghi sẵn kết quả.

| Phép toán | Toán tử | Kết quả mong đợi |
|---|---|---|
| Cộng | `+` | 22 |
| Trừ | `-` | 12 |
| Nhân | `*` | 85 |
| Chia | `/` | 3.4 |
| Chia lấy phần nguyên (với các số dương ở đây) | `//` | 3 |
| Chia lấy dư | `%` | 2 |
| Lũy thừa: a mũ b | `**` | 1419857 |

**Đoán thêm:** `(a + b) * 2` khác `a + b * 2` như thế nào?

````python
a = 17
b = 5

sum_value = a + b
difference = 4
product = 5
quotient = 6
whole_part = 7
remainder = 8
power_value = 9

print(sum_value, difference, product, quotient, whole_part, remainder, power_value)
````

````python
# Ô kiểm tra có sẵn — không cần sửa hoặc học cú pháp trong ô này.
if sum_value is None or difference is None or product is None or quotient is None or whole_part is None or remainder is None or power_value is None:
    print("✏️ Hãy hoàn thành các vị trí None ở bài phía trên.")
else:
    assert (sum_value, difference, product, quotient, whole_part, remainder, power_value) == (22, 12, 85, 3.4, 3, 2, 1419857), "Chưa đúng. Hãy xem lại yêu cầu và chạy lại ô bài làm."
    print("✅ Kết quả đúng!")
````

## 3 · Từ chuỗi đến số 
`input()` luôn trả về chuỗi, kể cả khi người dùng gõ một con số. Trước khi tính toán, ta chuyển chuỗi sang số bằng `int()` hoặc `float()`.

Ví dụ: `int("18")` → `18`; `float("8.5")` → `8.5`.

**Em làm:** hai chuỗi bên dưới mô phỏng dữ liệu nhận từ bàn phím. Chuyển chúng thành số thực rồi tính trung bình hai điểm. Kết quả mong đợi: `8.0`.

````python
score_1_text = "8.5"
score_2_text = "7.5"

score_1 = float(score_1_text)
score_2 = float(score_2_text)
average_score = (score_1 + score_2) /2

print("Điểm trung bình:", average_score)
````

````python
# Ô kiểm tra có sẵn — không cần sửa hoặc học cú pháp trong ô này.
if score_1 is None or score_2 is None or average_score is None:
    print("✏️ Hãy hoàn thành các vị trí None ở bài phía trên.")
else:
    assert type(score_1) is float and type(score_2) is float and score_1 == 8.5 and score_2 == 7.5 and average_score == 8.0, "Chưa đúng. Hãy xem lại yêu cầu và chạy lại ô bài làm."
    print("✅ Kết quả đúng!")
````

⚠️ `int("8.5")` gây lỗi vì chuỗi này biểu diễn số thập phân. Trong bài này, hãy dùng `float()` để đọc điểm.

## 4 · In kết quả gọn bằng f-string 
Đọc và chạy ví dụ:

````python
example_name = "Lan"
example_score = 7.0
print(f"{example_name}: {example_score:.2f} điểm")
````

`f"...{biến}..."` chèn giá trị vào chuỗi. `:.2f` hiển thị số với hai chữ số sau dấu thập phân; không đổi giá trị số ban đầu.

**Em làm:** dùng f-string tạo biến `message` từ hai biến bên dưới.

Chuỗi cần tạo chính xác: **`Minh: điểm trung bình = 8.25`**

````python
report_name = "Minh"
report_average = 8.25
message = f"{report_name}: diem trung binh = {report_average} diem"
print(message)
````

## 5 · Đổi phút thành giờ và phút 
**Bài toán:** có `185` phút. Hãy tính số giờ trọn vẹn và số phút còn lại.

Ví dụ: 125 phút → 2 giờ 5 phút. Với 185 phút, kết quả phải là **3 giờ 5 phút**.

Gợi ý: một giờ có 60 phút. Chọn hai toán tử phù hợp trong bài 2.

````python
total_minutes = 185
hours = total_minutes // 60
minutes_left = total_minutes % 60 

print(f"{total_minutes} phut doi ra gio la {hours},{minutes_left}h")
````

````python
# Ô kiểm tra có sẵn — không cần sửa hoặc học cú pháp trong ô này.
if hours is None or minutes_left is None:
    print("✏️ Hãy hoàn thành các vị trí None ở bài phía trên.")
else:
    assert hours == 3 and minutes_left == 5 and hours * 60 + minutes_left == total_minutes, "Chưa đúng. Hãy xem lại yêu cầu và chạy lại ô bài làm."
    print("✅ Kết quả đúng!")
````

**Tự thử:** đổi `total_minutes` lần lượt thành `0`, `59`, `60`, `125` và chạy lại **ô bài làm**. Đối chiếu bảng sau. Ô kiểm tra bên trên chỉ dành cho dữ liệu gốc `185`; hãy khôi phục giá trị này trước khi dùng nó.

| Tổng phút | Giờ | Phút còn lại |
|---:|---:|---:|
| 0 | 0 | 0 |
| 59 | 0 | 59 |
| 60 | 1 | 0 |
| 125 | 2 | 5 |

## 6 · Tự viết chương trình tính điểm 
Gộp các kiến thức vừa học thành một chương trình:

1. Nhập họ tên bằng `input()`.
2. Nhập điểm Toán và điểm Anh, chuyển sang `float`.
3. Tính trung bình hai điểm. Chú ý dấu ngoặc!
4. In kết quả bằng f-string với hai chữ số thập phân.

Ví dụ tương tác:
```text
Họ tên: Lan
Điểm Toán: 8
Điểm Anh: 6
Lan: điểm trung bình = 7.00
```

**Nhắc nhanh:** `name = input("Họ tên: ")`. Chương trình sẽ chờ em nhập dữ liệu khi chạy. Dùng dấu chấm cho số thập phân, ví dụ `8.5`. Tạm giả sử người dùng nhập điểm hợp lệ từ 0 đến 10; chưa cần xử lý nhập sai.

**Em tự viết toàn bộ chương trình trong ô dưới.**

````python
name = input("Ho ten: ")
Toan = int(input("Diem Toan: "))
Anh = int(input("Diem Anh: "))
dTB = (Toan + Anh) /2
print(f"{name}: diem trung binh {dTB}")
````
